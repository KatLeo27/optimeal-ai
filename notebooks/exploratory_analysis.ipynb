## 2. Processed & Augmented Dataset Overview

```python
proc_df = load_processed_data('../data/processed/food_waste_processed.csv')
print(f'Processed dataset shape: {proc_df.shape}')
proc_df[['Date', 'Meal', 'Food_Category', 'Waste_Weight_kg', 'Cost_Loss', 'expected_attendance', 'meals_consumed', 'surplus_meals']].head()
```

## 3. Exploratory Data Visualizations

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Waste weight by category
sns.barplot(data=proc_df, x='Food_Category', y='Waste_Weight_kg', ax=axes[0], palette='crest')
axes[0].set_title('Average Waste Weight (kg) by Category', fontsize=12, fontweight='bold')
axes[0].set_ylabel('Waste Weight (kg)')

# Cost loss by category
sns.barplot(data=proc_df, x='Food_Category', y='Cost_Loss', ax=axes[1], palette='flare')
axes[1].set_title('Average Financial Cost Loss ($) by Category', fontsize=12, fontweight='bold')
axes[1].set_ylabel('Cost Loss ($)')

plt.tight_layout()
plt.show()
```

## 4. Attendance vs Meals Consumed Correlation

```python
plt.figure(figsize=(8, 6))
sns.scatterplot(data=proc_df, x='expected_attendance', y='meals_consumed', hue='Meal', alpha=0.7, palette='tab10')
plt.title('Expected Attendance vs Actual Meals Consumed', fontsize=13, fontweight='bold')
plt.xlabel('Expected Attendance (Headcount)')
plt.ylabel('Meals Consumed (Portions)')
plt.show()
```
